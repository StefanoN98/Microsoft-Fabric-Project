```python
import requests
import json
import notebookutils
```

```python
# 1. Set Fabric parameters
workspace_id = "47f4b971-7b6b-4af9-8001-77745ea917c0"
lakehouse_id = "c9c219f7-14cc-47f2-8994-35d08101e937"  # LH_Bronze
connection_id = "93ab585b-2e1f-4f61-b87d-a77ff0beb7b5" # connection ID between ADLS Gen 2 and Fabric

# 2. Shortcut details to create
shortcut_name = "shortcut_adls"
adls_location = "https://stglobalretail.dfs.core.windows.net" # Endpoint DFS (not blob)
adls_subpath = "/landing/erp/Shortcut"         
target_path = "Files" 
```

```python
# 3. Automatic generation for authentication token
# notebookutils retrieves the token of the logged user, without necessity to use password in the notebook
token = notebookutils.credentials.getToken("pbi")

# 4. Define endpoint for API REST in Fabric
url = f"https://api.fabric.microsoft.com/v1/workspaces/{workspace_id}/items/{lakehouse_id}/shortcuts"

# 5. Request payload
payload = {
    "name": shortcut_name,
    "path": target_path,
    "target": {
        "adlsGen2": {
            "location": adls_location,
            "subpath": adls_subpath,
            "connectionId": connection_id
        }
    }
}

headers = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}
```

```python
# 6. Send creation request
response = requests.post(url, headers=headers, json=payload)

if response.status_code == 201:
    print(f"✅ Shortcut '{shortcut_name}' creato con successo in {target_path} all'interno di LH_Bronze!")
else:
    print(f"❌ Errore durante la creazione del shortcut: {response.status_code}")
    print(response.text)
```
