```python
import urllib.request
import os

# Create folders if it doesn't exist
os.makedirs("/lakehouse/default/Files/ecommerce", exist_ok=True)

# Direct dowloand using Github URL  and solve Git LFS
lfs_url = "https://github.com/StefanoN98/Microsoft-Fabric-Project/raw/main/globalretail-source-data/ecommerce/web_logs.json"
destination_path = "/lakehouse/default/Files/ecommerce/web_logs.json"

print("Download del file grande web_logs.json in corso (286MB)...")
try:
    urllib.request.urlretrieve(lfs_url, destination_path)
    print("✅ Download completato con successo!")
except Exception as e:
    print(f"❌ Errore durante il download: {e}")
```
